```python

import pandas as pd
import numpy as np
from pathlib import Path

# Locate the climate dataset
climate_path = Path("data/raw/nasa_climate_monthly.csv")

climate = pd.read_csv(climate_path)

print("Original shape:", climate.shape)
print("Columns:", climate.columns.tolist())

# Standardize column names
climate.columns = climate.columns.str.strip().str.lower()

# Verify required columns
required = [
    "country",
    "iso3",
    "date",
    "temperature_c",
    "precipitation_mm"
]

missing_columns = set(required) - set(climate.columns)

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

# Keep the required fields
climate = climate[required].copy()

# Convert sentinel values to missing values
numeric_columns = [
    "temperature_c",
    "precipitation_mm"
]

for col in numeric_columns:
    climate[col] = pd.to_numeric(
        climate[col],
        errors="coerce"
    )

    climate.loc[
        climate[col] == -999,
        col
    ] = np.nan

# Parse YYYYMM dates
date_string = (
    climate["date"]
    .astype("string")
    .str.replace(r"\.0$", "", regex=True)
    .str.strip()
)

climate["date"] = pd.to_datetime(
    date_string,
    format="%Y%m",
    errors="coerce"
)

# Flag implausible measurements for review
# These are quality-control bounds, not universal climate limits.
climate.loc[
    ~climate["temperature_c"].between(-90, 65),
    "temperature_c"
] = np.nan

climate.loc[
    climate["precipitation_mm"] < 0,
    "precipitation_mm"
] = np.nan

# Remove records without valid identifiers or dates
climate = climate.dropna(
    subset=["country", "iso3", "date"]
)

# Check duplicate country-month observations
duplicates = climate.duplicated(
    subset=["iso3", "date"],
    keep=False
)

print("\nDuplicate country-month rows:", duplicates.sum())

if duplicates.any():
    display(
        climate.loc[duplicates]
        .sort_values(["iso3", "date"])
        .head(20)
    )

# Do not silently aggregate duplicates.
# Investigate their source before deciding how to handle them.

# Sort records
climate = climate.sort_values(
    ["iso3", "date"]
).reset_index(drop=True)

print("\nCleaned shape:", climate.shape)

print("\nMissing values:")
print(climate.isna().sum())

print("\nCleaned summary statistics:")
display(climate[numeric_columns].describe())

# Save the cleaned dataset
output_path = Path("data/processed")
output_path.mkdir(parents=True, exist_ok=True)

cleaned_file = output_path / "climate_monthly_clean.csv"

climate.to_csv(cleaned_file, index=False)

print("\nSaved:", cleaned_file)
```

```python

def climate_signal(temp_anomaly=None, precip_anomaly=None):
    vals = []

    if pd.notna(temp_anomaly):
        vals.append(np.clip(abs(temp_anomaly) / 3, 0, 1))

    if pd.notna(precip_anomaly):
        vals.append(np.clip(abs(precip_anomaly) / 100, 0, 1))

    return float(np.mean(vals)) if vals else 0.0
```

```python

import pandas as pd
import numpy as np

climate = pd.read_csv(
    "data/processed/climate_monthly_clean.csv",
    parse_dates=["date"]
)

# Derive calendar month
climate["month"] = climate["date"].dt.month

# Choose a reference period with sufficient historical coverage.
# This example uses 2015–2024 because the dataset begins in 2015.
baseline = climate[
    climate["date"].dt.year.between(2015, 2024)
].copy()

# Calculate each country's typical value for each calendar month
monthly_baseline = (
    baseline
    .groupby(["iso3", "month"], as_index=False)
    .agg(
        temp_baseline_c=("temperature_c", "mean"),
        precip_baseline_mm=("precipitation_mm", "mean"),
        temp_baseline_n=("temperature_c", "count"),
        precip_baseline_n=("precipitation_mm", "count")
    )
)

# Join the historical baselines to the observations
climate = climate.merge(
    monthly_baseline,
    on=["iso3", "month"],
    how="left",
    validate="many_to_one"
)

# Absolute temperature anomaly, in degrees Celsius
climate["temp_anomaly_c"] = (
    climate["temperature_c"]
    - climate["temp_baseline_c"]
)

# Percentage precipitation anomaly
# Suppress percentages when the baseline is zero or unavailable.
climate["precip_anomaly_pct"] = np.where(
    climate["precip_baseline_mm"] > 0,
    (
        (
            climate["precipitation_mm"]
            - climate["precip_baseline_mm"]
        )
        / climate["precip_baseline_mm"]
    ) * 100,
    np.nan
)

# Do not claim an anomaly where the historical reference
# has insufficient observations.
climate.loc[
    climate["temp_baseline_n"] < 5,
    "temp_anomaly_c"
] = np.nan

climate.loc[
    climate["precip_baseline_n"] < 5,
    "precip_anomaly_pct"
] = np.nan

print("Temperature anomaly summary:")
display(climate["temp_anomaly_c"].describe())

print("\nPrecipitation anomaly summary:")
display(climate["precip_anomaly_pct"].describe())

print("\nSample records:")
display(
    climate[
        [
            "country",
            "date",
            "temperature_c",
            "temp_baseline_c",
            "temp_anomaly_c",
            "precipitation_mm",
            "precip_baseline_mm",
            "precip_anomaly_pct"
        ]
    ].tail(15)
)

climate.to_csv(
    "data/processed/climate_anomalies.csv",
    index=False
)

print("\nSaved climate anomaly dataset.")
```

```python

import numpy as np
import pandas as pd
from pathlib import Path

# Load the climate anomalies calculated in Step 3
climate = pd.read_csv("data/processed/climate_anomalies.csv")

def calculate_climate_component(row):
    """
    Calculate a bounded, exploratory climate signal from available
    temperature and precipitation anomalies.

    This is a portfolio screening heuristic, not a validated
    disease-outbreak prediction model.
    """
    signals = []

    # Temperature anomaly: absolute deviation in degrees Celsius
    temp = row["temp_anomaly_c"]

    if pd.notna(temp):
        signals.append(float(np.clip(abs(temp) / 3.0, 0, 1)))

    # Precipitation anomaly: percentage deviation from baseline
    precip = row["precip_anomaly_pct"]

    if pd.notna(precip):
        signals.append(float(np.clip(abs(precip) / 100.0, 0, 1)))

    # Do not interpret missing climate observations as zero risk
    if not signals:
        return np.nan

    return float(np.mean(signals))


# Calculate the climate component for each country-month
climate["climate_component"] = climate.apply(
    calculate_climate_component,
    axis=1
)

# Validate the calculation
print("Climate component calculated.")
print("Total records:", len(climate))
print("Missing climate components:", climate["climate_component"].isna().sum())

display(
    climate[
        [
            "country",
            "date",
            "temp_anomaly_c",
            "precip_anomaly_pct",
            "climate_component"
        ]
    ].head(15)
)

# Inspect the distribution
display(climate["climate_component"].describe())

# Save the validated intermediate output
output_path = Path("data/processed/climate_anomalies.csv")
climate.to_csv(output_path, index=False)

print(f"Updated file saved to: {output_path}")
```

```python

from pathlib import Path
import ast

path = Path("src/risk_engine.py")
source = path.read_text()

def replace_function(source, function_name, replacement):
    tree = ast.parse(source)

    matches = [
        node for node in tree.body
        if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef))
        and node.name == function_name
    ]

    if len(matches) != 1:
        raise ValueError(
            f"Expected exactly one {function_name}() function; "
            f"found {len(matches)}."
        )

    node = matches[0]
    return (
        source[:node.col_offset + node.lineno * 0
               + sum(len(line) for line in source.splitlines(keepends=True)[:node.lineno - 1])]
        + replacement.rstrip() + "\n\n"
        + source[
            sum(len(line) for line in source.splitlines(keepends=True)[:node.end_lineno]):
        ]
    )

vaccination_function = '''def vaccination_gap(vax_percent):
    """Return a normalized vaccination gap, preserving missing values."""
    if pd.isna(vax_percent):
        return np.nan
    return float(np.clip((95 - vax_percent) / 50, 0, 1))'''

climate_function = '''def climate_signal(temp_anomaly=None, precip_anomaly=None):
    """Summarize available climate anomalies; preserve missing observations."""
    vals = []

    if pd.notna(temp_anomaly):
        vals.append(np.clip(abs(temp_anomaly) / 3, 0, 1))

    if pd.notna(precip_anomaly):
        vals.append(np.clip(abs(precip_anomaly) / 100, 0, 1))

    return float(np.mean(vals)) if vals else np.nan'''

updated = replace_function(source, "vaccination_gap", vaccination_function)
updated = replace_function(updated, "climate_signal", climate_function)

# Verify syntax before writing the file.
ast.parse(updated)

path.write_text(updated)
print("PASS: Updated vaccination_gap() and climate_signal().")
print("PASS: Python syntax is valid.")
```

```python

%cd /content/africa-public-health-intelligence

from pathlib import Path
import re

path = Path("src/pipeline.py")
source = path.read_text(encoding="utf-8")

new_function = '''def forecast(years, values, current_year=2026):
    """
    Forecast the next annual disease value conservatively.

    A forecast is returned only when:
    - the latest observation is no more than two years old;
    - at least six observations are available;
    - the six most recent observations cover consecutive calendar years.

    Missing years are not treated as zero.
    """
    data = pd.DataFrame({
        "year": pd.to_numeric(pd.Series(years), errors="coerce"),
        "value": pd.to_numeric(pd.Series(values), errors="coerce"),
    })

    data = data.dropna(subset=["year", "value"]).copy()
    data = data.sort_values("year")

    if data.empty:
        return np.nan

    # Annual surveillance records should have integer calendar years.
    data = data[data["year"] % 1 == 0].copy()
    data["year"] = data["year"].astype(int)

    # Multiple observations for one country-disease-year are ambiguous.
    if data["year"].duplicated().any():
        return np.nan

    latest_year = int(data["year"].iloc[-1])

    # Do not present old surveillance data as a current forecast.
    if current_year - latest_year > 2 or latest_year > current_year:
        return np.nan

    recent = data.tail(6).copy()

    if len(recent) < 6:
        return np.nan

    # Require consecutive calendar years in the six observations.
    year_differences = recent["year"].diff().dropna()

    if not year_differences.eq(1).all():
        return np.nan

    s = recent["value"].astype("float64")

    if not np.isfinite(s).all() or (s < 0).any():
        return np.nan

    try:
        model = ExponentialSmoothing(
            s.reset_index(drop=True),
            trend="add",
            damped_trend=True,
            initialization_method="estimated",
        ).fit(optimized=True)

        forecast_value = float(model.forecast(1).iloc[0])

        if np.isfinite(forecast_value):
            return max(0.0, forecast_value)

    except (
        ValueError,
        TypeError,
        ArithmeticError,
        np.linalg.LinAlgError,
    ):
        pass

    # Conservative fallback for a valid recent annual series.
    fallback = float(s.tail(3).median())

    return max(0.0, fallback) if np.isfinite(fallback) else np.nan
'''

pattern = r"def forecast\(series\):.*?(?=\ndef build\(\):)"

updated_source, replacements = re.subn(
    pattern,
    new_function.rstrip() + "\n\n",
    source,
    count=1,
    flags=re.DOTALL,
)

if replacements != 1:
    raise RuntimeError(
        f"Expected to replace one forecast function; replaced {replacements}. "
        "The file has not been changed."
    )

# Write only after the expected function has been found.
path.write_text(updated_source, encoding="utf-8")

print("Forecast function replaced successfully.")
print("\n".join(
    f"{i + 1:03d}: {line}"
    for i, line in enumerate(updated_source.splitlines()[135:205], start=135)
))
```
